```python
import numpy as np
```

```python
#3.1 
numpy_pi = np.pi 
print(numpy_pi)
```

```python
#3.2
def series_pi(n=1000):
    series_pi = 0
    for k in range(n):
        term = (-1)**k / (2*k+1)
        series_pi += term
    series_pi = 4 * series_pi
    return series_pi
```

```python
#3.3 
def mc_pi(n=1000):
    rand_xy = np.random.uniform(size=(n, 2))
    
    is_in_cycle = np.sqrt(rand_xy[:, 0]**2 + rand_xy[:, 1]**2) < 1
    n_circle = is_in_cycle.sum()
    mc_pi = 4 * n_circle/n
    return mc_pi
```

```python
n = 1000
s_pis = []
mc_pis = []
for i in range(10):
    s_pis.append(series_pi(n))
    mc_pis.append(mc_pi(n))

print(np.array(s_pis).var())
print(np.array(mc_pis).var())
```

```python
import matplotlib.pyplot as plt

plt.scatter(np.arange(len(s_pis)), s_pis)
```
