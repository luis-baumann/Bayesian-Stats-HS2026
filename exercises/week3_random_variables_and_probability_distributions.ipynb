### Exercise 1
Consider the distribution with PDF
$$
    p(x) = \begin{cases}
                \frac{1}{2}\sin x & 0 \le x \le \pi \\
                0 & \text{otherwise}
            \end{cases}
$$
1. Derive the CDF $F(x) = \int_{-\infty}^{x} p(x')\,\dd x'$ analytically, and plot it together with the PDF $p(x)$.
2. Find the inverse CDF $F^{-1}(u)$, and use it to transform samples $U \sim \Uniform(0, 1)$ into samples $X = F^{-1}(U)$ (inverse transform sampling).
3. Compare a density-normalised histogram of the samples with the PDF $p(x)$.

```python
import numpy as np
import matplotlib.pyplot as plt


def pdf(x):
    return np.where(((x < 0) | (x > np.pi)), 0, 0.5 * np.sin(x))


    
def cdf(x):
    condlist = [x < 0, x > np.pi]
    choicelist = [0, 1]
    return np.select(condlist, choicelist, default=(0.5*(1-np.cos(x))))

def inverse_cdf(u):
    return np.arccos(1-2*u)

def inv_cdf_sample(n=100000000):
    rand = np.random.rand(n)
    return inverse_cdf(rand)

x = np.linspace(-1, np.pi+1, 100)
u = np.linspace(0, 1, 100)
pdf_line = plt.plot(x, pdf(x))[0]
plt.plot(x, cdf(x))
plt.hist(inv_cdf_sample(), bins=1000, density=True, alpha=0.5, color=pdf_line.get_color())
plt.show()
```

### Exercise 3

Let $X_1, \dots, X_n$ be $n$ i.i.d. random variables with $X_i \sim \Uniform(0, 1)$, and let $K$ be the number of them that fall into a small interval of width $\Delta x$.

1. Confirm by simulation that, for large $n$ and small $\Delta x$ with the rate $\lambda = n\,\Delta x$ held fixed, $K$ follows a Poisson distribution:
$$
    \Pr(K = k) = \frac{\lambda^k}{k!}\ee^{-\lambda}
$$
2. Derive the Poisson distribution from the binomial distribution, $\Pr(K = k) = \binom{n}{k}p^k(1-p)^{n-k}$, in the limit $n \to \infty$ with $np = \lambda$ held fixed.

```python
import math
n = 100
dx = 0.05

vals = np.random.rand(10000, n)

interval = np.where(vals < dx, 1, 0)
res = np.sum(interval, axis=1)

ks = np.bincount(res, minlength=n+1) / n

lam = n*dx 
def theo_pr(k):
    return lam**k / math.factorial(k) * np.exp(-lam)

plt.scatter(np.arange(len(ks+1)), ks, s=5)

k_th_vals = np.arange(len(ks+1)).astype('int')
print(k_th_vals)
plt.scatter(k_th_vals, theo_pr(k_th_vals))
plt.xlim(0, 20)
```
